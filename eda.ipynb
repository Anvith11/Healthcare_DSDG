## Target Variables Analysis

### Primary Target
- **Health Expenditure % of GDP** — measures how much of the economy goes to health; consistent target across all regions and the strongest outcome variable for cross-country comparison.

### Secondary Targets

- **Govt Health Expenditure % of GDP** — measures government's direct health investment; the single strongest predictor of the primary target across all regions (r = 0.757–0.865), making it both a feature and a viable target depending on the modelling angle.

- **Out-of-Pocket Share of Health Expenditure (%)** — key measure of financial burden on households; directly actionable by policy and strongly driven by government coverage levels (r = -0.717 to -0.920 across regions).

- **Private Share of Health Expenditure (%)** — reflects how much falls outside public coverage; near-perfectly inversely correlated with Govt Share across all regions (-0.853 to -0.995), making it a direct proxy for coverage gaps.

- **Govt Share of Health Expenditure (%)** — measures public coverage breadth; core policy lever and the most consistently correlated variable with OOP burden across all regions.

- **General Govt Expenditure % of GDP** — broader fiscal commitment to public services; upstream driver of health spending, moderately correlated with the primary target across all regions (r = 0.494–0.695).

### Target Selection by Research Goal

- **Reducing total health spend** → use `Health Expenditure % of GDP`
- **Reducing household financial burden** → use `Out-of-Pocket Share of Health Expenditure (%)`
- **Optimising government allocation** → use `Govt Health Expenditure % of GDP`
- **Expanding population coverage** → use `Govt Share of Health Expenditure (%)`
